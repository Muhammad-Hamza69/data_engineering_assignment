Q3: Write a program in which use all the operators we can use in Python

```python
num_1 = 10
num_2 = 3

print("the addition is: ",num_1 + num_2)
print("the subtraction is: ",num_1 - num_2)
print("the multiplication is: ",num_1 * num_2)
print("the division is: ",num_1 / num_2)
print("the floor division is: ",num_1 // num_2)
print("the modulus is: ",num_1 % num_2)
print("the power is: ",num_1 ** num_2)
print("the greater than is: ",num_1 > num_2)
print("the less than is: ",num_1 < num_2)
print("the greater than equal to is: ",num_1 >= num_2)
print("the less than equal to is: ",num_1 <= num_2) 
print("the equal is: ",num_1 == num_2)
print("the not equal is: ",num_1 != num_2)
print("the and of two number is: ",num_1 and num_2)
print("the or of two number is: ",num_1 or num_2)
print("the not of two operator is: ",not(num_1 and num_2))
num_1+=num_2
print("the assignment operator with positive is: ",num_1)
num_1-=num_2
print("the assignment operator with negative is:  ",num_1)
num_1*=num_2
print("the the assignment operator with multiplication is:  ",num_1)
num_1/=num_2
print("the the assignment operator with division is:  ",num_1)
num_1**num_2
print("the the assignment operator with division is:  ",num_1)
num_1//=num_2
print("the the assignment operator with division is:  ",num_1)

print("the identity operator is: ", num_1 is num_2)

```
