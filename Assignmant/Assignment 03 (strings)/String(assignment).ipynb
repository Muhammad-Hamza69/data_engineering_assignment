Q1: Write a program that accepts string from user. Your program should count and display number of vowels in that string.

```python
user_input = input("enter a any message: ")
count = 0
for x in user_input:
    if (x == 'a'):
        count+=1
    elif(x == 'e'):
        count+=1
    elif(x == 'i'):
        count+=1
    elif(x == 'o'):
        count+=1
    elif(x == 'u'):
        count+=1
if(count): 
    print(f"the number of vowels in your string are: {count}")        
else:
    print(f"your string have {count} vowels")
    
```

Q2: Write a program that reads a string from keyboard and display: 
    * The number of Uppercase letter in the string
    * the number of lowercase letter in the string
    * the number of digits in the string
    * the number of whitecase characters in the string

```python
user_input = input("enter a any message: ")

uppercase_count = 0
lowercase_count = 0
digit_count = 0
whitecase_count = 0

for x in user_input:
    print(x)
    if(x.isupper()):
        uppercase_count+=1
    elif(x.islower()):
        lowercase_count+=1
    elif(x.isdigit()):
        digit_count+=1  
    elif(x.isspace()):
        whitecase_count+=1    
print(f"the number of uppercase character in your message are: {uppercase_count}")
print(f"the number of lowercase character in your message are: {lowercase_count}")
print(f"the number of digit character in your message are: {digit_count}")
print(f"the number of whitecase character in your message are: {whitecase_count}")

        
    
    
```

Q3: write a python program that accepts a string from user. Your program should create and display a new string where the first and last character have been exchanged 

```python
user_input = input("enter a any message: ")
r=""    
for x in range(len(user_input)):
    if(x==0):
        r = r + 'O'
    elif(x == (len(user_input)-1)):
        r = r + 'H'
    else:
        r = r + user_input[x]

print(r)
        
    
```

Q9: write a program in python that accept a string to setup a passwords. Your entered password must meet the following requirements:   
- the password must be at least eight characters long.
- it must contain at least one uppercase letter
- it must contain at least one lowercase letter
- it must contain at least one numeric digit
- your program should perform this validation

```python
password = input("enter any message: ")
upper = False
lower = False
digit = False
if(len(password) >= 8):
    for x in password:
        if(x.isupper()):
            upper = True
        elif(x.islower()):
            lower = True
        elif(x.isdigit()):
            digit = True
    if(upper and lower and digit):
        print("your password is valid")
    else:
        print("your password is not accepted, please enter the valid password")
        

else:
    print("your password is not accepted, please enter the valid password")
            
    
```
