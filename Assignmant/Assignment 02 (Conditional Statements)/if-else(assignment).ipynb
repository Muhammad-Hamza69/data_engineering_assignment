Q: A company decided to give bonus of 5% to employee if his/her year of service is more than 5 years. Ask user for their salary and year of service and print the net bonus amount.

```python
import math 
current_salary = int(input("please enter your salary: "))
year_of_services = int(input("now, enter your year of service: "))

if year_of_services>5:
    current_salary = math.trunc(current_salary + ((5/100) * current_salary))
    print(f"congratulation you have acheived 5% bonus on your current salary: {current_salary}")
    
```

Q: A shop will give discount of 10% if the cost of purchased quantity is more than 1000. Ask user for quantity Suppose, one unit will cost 100. Judge and print total cost for user.

```python
quantity = int(input("enter the quantity you have purchased: "))
cost_of_one_unit = 100
total_cost = math.trunc(quantity * cost_of_one_unit)

if total_cost>1000:
    total_cost = math.trunc(total_cost - ((10/100) * total_cost))
    print(f"you have get a 10% discount so the total cost you have to paid is {total_cost}")
else:
    print(f"so the total cost you have to paid is {total_cost}")
    
```

Q: A student will not be allowed to sit in exam if his/her attendence is less than 75%.

Take following input from user

Number of classes held

Number of classes attended.

And print

percentage of class attended

Is student is allowed to sit in exam or not.

```python
Number_of_classes_held = int(input("enter the number of classes held: "))
Number_of_classes_attended = int(input("enter the number of classes attended: "))

percentage = math.trunc((Number_of_classes_attended / Number_of_classes_held) * 100)
print(percentage)

if percentage<75:
    print("you are not allowed to sit in the exam")
else:
    print("you are allowed to sit in the exam")
```

Q: Write a program to check if a year is leap year or not.

```python
year = int(input("enter any year: "))

if ((year%4==0 and year%100!=0) or year%400==0):
    print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
    
```
